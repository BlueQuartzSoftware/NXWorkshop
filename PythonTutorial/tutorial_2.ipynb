## Historical Method of Iterating on Pipelines

- Create the pipeline
- Open a text editor and replace inputs with easily found key words
- Save pipeline file
- Create a script that will:
- Read the pipeline as a string into memory
- Do text replacements
- Save to another pipeline file
- Instantiate "PipelineRunner" to execute the pipeline
- This is still entirely possible but maybe not the best for every situation


# Slide 4: Opening/Modifying Pipelines in Python

### Reading and Writing a Pipeline File
- `Pipeline.from_file(path)`
- `Pipeline.to_file(path)`

### Pipeline API

A pipeline is essentially a list of filters so the API should be familiar to the native Python `list`. 

- `insert`
- `append`
- `clear`
- `remove`
- `size`


# Slide 5: Example Pipeline

![](Images/Tutorial_2/Slide_5.png)

```python
import simplnx as nx
# These imports are required despite not being used to load filters used in the pipeline here.
# See note at the end.
import itkimageprocessing as nxitk
import orientationanalysis as nxor
# This is a utility function for printing warnings/errors from executing a pipeline.
# Then if there is an error, it will raise an exception.
from nxutility import check_pipeline_result
# Convenience function to easily modify a specific filter in the pipeline.
# modify_pipeline_filter(filter: nx.PipelineFilter, key: str, value)
# This is needed until the next release of DREAM3D-NX.
from nxutility import modify_pipeline_filter
# See PythonTutorial/nxutility.py for details
from IPython.display import Image
```

```python
# Initial load

INPUT_FILE_KEY = 'input_file'
FILE_NAME_KEY = 'file_name'
EXPORT_FILE_PATH_KEY = 'export_file_path'

# Load the Pipeline from file
pipeline = nx.Pipeline.from_file('lesson_2_ebsd.d3dpipeline')

# Acquire filters that we want to modify
read_ang_data_filter: nx.PipelineFilter = pipeline[0]
itk_image_writer_filter: nx.PipelineFilter = pipeline[5]
write_dream3d_filter: nx.PipelineFilter = pipeline[6]
```

```python
original_output_image_path = 'Output/Small_IN100/Slice_1.png'

# Modifiy pipeline to fix file paths
modify_pipeline_filter(read_ang_data_filter, INPUT_FILE_KEY, 'Data/Small_IN100/Slice_1.ang')

modify_pipeline_filter(itk_image_writer_filter, FILE_NAME_KEY, original_output_image_path)

modify_pipeline_filter(write_dream3d_filter, EXPORT_FILE_PATH_KEY, 'Output/Small_IN100/Slice_1.dream3d')

# Run pipeline
original_data_structure = nx.DataStructure()
original_result = pipeline.execute(original_data_structure)
check_pipeline_result(original_result)

# Display result
Image(filename=original_output_image_path)
```

```python
modified_output_image_path = 'Output/Small_IN100/Slice_5.png'

# Modify input and output file paths
modify_pipeline_filter(read_ang_data_filter, INPUT_FILE_KEY, 'Data/Small_IN100/Slice_5.ang')

modify_pipeline_filter(itk_image_writer_filter, FILE_NAME_KEY, modified_output_image_path)

modify_pipeline_filter(write_dream3d_filter, EXPORT_FILE_PATH_KEY, 'Output/Small_IN100/Slice_5.dream3d')

# Always need a DataStructure Object
modified_data_structure = nx.DataStructure()

# Execute the Pipelne
modified_result = pipeline.execute(modified_data_structure)
check_pipeline_result(modified_result)

# Save the modified pipeline to file
pipeline.to_file('Output/lesson_2_ebsd_modified.d3dpipeline')

Image(filename=modified_output_image_path)
```

# Important note on loading pipelines

Filters belong to individual plugins which must be loaded when importing a pipeline.

`import simplnx` will ensure that the core plugin and its filters are loaded.

The import statements `import itkimageprocessing` and `import orientationanalysis` will ensure that their respective filters will be available even if the Python modules themselves aren't used.
