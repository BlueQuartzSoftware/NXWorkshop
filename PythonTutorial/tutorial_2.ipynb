## Historical Method of Iterating on Pipelines

- Create the pipeline
- Open a text editor and replace inputs with easily found key words
- Save pipeline file
- Create a script that will:
- Read the pipeline as a string into memory
- Do text replacements
- Save to another pipeline file
- Instantiate "PipelineRunner" to execute the pipeline
- This is still entirely possible but maybe not the best for every situation


# Slide 4: Opening/Modifying Pipelines in Python

### Reading and Writing a Pipeline File
- `Pipeline.from_file(path)`
- `Pipeline.to_file(path)`

### Pipeline API

A pipeline is essentially a list of filters so the API should be familiar to the native Python `list`. 

- `insert`
- `append`
- `clear`
- `remove`
- `size`


# Slide 5: Example Pipeline

![](Images/Tutorial_2/Slide_5.png)

```python
# Load the Pipeline from file
pipeline = nx.Pipeline.from_file('lesson_2_ebsd.d3dpipeline')

# Modify 3 of the filters within the pipeline
modify_pipeline_filter(pipeline[0], 'input_file', 'Data/Small_IN100/Slice_1.ang')

modify_pipeline_filter(pipeline[5], 'file_name', 'Output/Small_IN100/Slice_1.png')

modify_pipeline_filter(pipeline[6], 'export_file_path', 'Output/Small_IN100/Slice_1.dream3d')

# Always need a DataStructure Object
data_structure = nx.DataStructure()

# Execute the Pipelne
result = pipeline.execute(data_structure)

# Print any erorrs with the Pipeline execution
check_pipeline_result(result)

# Save the modified pipeline to file
pipeline.to_file('Output/lesson_2_ebsd_modified.d3dpipeline')
```

# Important note on loading pipelines

Filters belong to individual plugins which must be loaded when importing a pipeline.

`import simplnx` will ensure that the core plugin and its filters are loaded.

The import statements `import itkimageprocessing` and `import orientationanalysis` will ensure that their respective filters will be available even if the Python modules themselves aren't used.
