```python
import simplnx as nx
import itkimageprocessing as nxitk
import orientationanalysis as nxor

def check_pipeline_result(result: nx.Result) -> None:
  """
  This function will check the `result` for any errors. If errors do exist then a 
  `RuntimeError` will be thrown. Your own code to modify this to return something
  else that doesn't just stop your script in its tracks.
  """
  if len(result.warnings) != 0:
    for w in result.warnings:
      print(f'Warning: ({w.code}) {w.message}')
  
  has_errors = len(result.errors) != 0 
  if has_errors:
    #print(f'Pipeline :: Errors: {result.errors}')
    for err in result.errors:
      print(f'Error: ({err.code}) {err.message}')
    raise RuntimeError(result)
  
  print("Pipeline :: No errors running the pipeline")
```

```python
# This is needed until the next release of DREAM3D-NX later on
# Easily modify a specific filter in the pipeline
def modify_pipeline_filter(filter: nx.PipelineFilter, key: str, value):
  param_dict = filter.get_args()
  param_dict[key] = value
  filter.set_args(param_dict)

# Load the Pipeline from file
pipeline = nx.Pipeline.from_file('lesson_2_ebsd.d3dpipeline')

# Modify 3 of the filters within the pipeline
modify_pipeline_filter(pipeline[0], "input_file", "Data/Small_IN100/Slice_1.ang")

modify_pipeline_filter(pipeline[5], "file_name", "Output/Small_IN100/Slice_1.png")

modify_pipeline_filter(pipeline[6], "export_file_path","Output/Small_IN100/Slice_1.dream3d")

# Always need a DataStructure Object
data_structure = nx.DataStructure()

# Execute the Pipelne
result = pipeline.execute(data_structure)

# Print any erorrs with the Pipeline execution
check_pipeline_result(result=result)

# Save the modified pipeline to file
pipeline.to_file('Output/lesson_2_ebsd_modified.d3dpipeline')
```
